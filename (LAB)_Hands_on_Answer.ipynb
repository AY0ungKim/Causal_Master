```python
# Bring the 'Earth Engine' toolbox into this script.
import ee
# Bring the 'Time' tool to help the computer wait during tasks.
import time
```

```python
project_id = 'gee-example-483611' # Your Google Cloud Project ID
folder_name = "exercise" # The folder name in your Google Drive (customize)
region_asset = 'projects/gee-example-483611/assets/grid_5km_seoul' # Your uploaded shapefile (asset)
```

```python
try:
    # Try to initialize with the specific project ID
    ee.Initialize(project=project_id)
except Exception as e:
    # If initialization fails, run authentication workflow
    ee.Authenticate()
    ee.Initialize(project=project_id)
```

## 1. Snow cover in Copernicus Global Land Cover Layers dataset in 2019-01-01

```python
dataset = ee.ImageCollection("COPERNICUS/Landcover/100m/Proba-V-C3/Global")
region = ee.FeatureCollection(region_asset)

target_bands = ['tree-coverfraction']
filtered_coll = dataset.filterDate('2019-01-01','2020-01-01').select(target_bands)

def extract_stats(image):
    stats = image.reduceRegions(
        collection=region,
        reducer=ee.Reducer.mean(),
        scale=10
    )

    year = image.date().get('year')

    return stats.map(lambda f: f.set({
        'year': year
    }))

final_features = filtered_coll.map(extract_stats).flatten()
final_features = final_features.filter(ee.Filter.notNull(['mean']))

csv_columns = ['grid_id', 'year', 'mean']

description = "Exercise_M1_TREE"
task = ee.batch.Export.table.toDrive(
    collection=final_features,
    description=description,
    folder=folder_name,
    fileNamePrefix=description,
    fileFormat="CSV",
    selectors=csv_columns
)

task.start()
print(f"Export started! Please check your GEE Tasks or Google Drive folder: {folder_name}")
```

## 2. Daily Enhanced vegetation index (EVI) in MODIS dataset in 2019-08

```python
dataset = ee.ImageCollection("MODIS/MOD09GA_006_EVI")
region = ee.FeatureCollection(region_asset)

target_bands = ['EVI']
filtered_coll = dataset.filterDate('2019-08-01','2019-09-01').select(target_bands)

def extract_stats(image):
    stats = image.reduceRegions(
        collection=region,
        reducer=ee.Reducer.mean(),
        scale=10
    )

    year = image.date().get('year')
    month = image.date().get('month')
    day = image.date().get('day')

    return stats.map(lambda f: f.set({
        'year': year,
        'month': month,
        'day': day
    }))

final_features = filtered_coll.map(extract_stats).flatten()
final_features = final_features.filter(ee.Filter.notNull(['mean']))

csv_columns = ['grid_id', 'year', 'month', 'day', 'mean']

description = "Exercise_M2_EVI"
task = ee.batch.Export.table.toDrive(
    collection=final_features,
    description=description,
    folder=folder_name,
    fileNamePrefix=description,
    fileFormat="CSV",
    selectors=csv_columns
)

task.start()
print(f"Export started! Please check your GEE Tasks or Google Drive folder: {folder_name}")
```
